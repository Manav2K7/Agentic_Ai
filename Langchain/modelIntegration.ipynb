```python
from langchain_groq import ChatGroq

model = ChatGroq(model_name="llama-3.1-8b-instant")
response = model.invoke("Hello How are you?")
response.content
```

Streaming and batch


```python
model.invoke("write me a 200 words oaragraph on LLM")
```

```python
for chunk in model.stream("Hey write me a 200 words paragraph on LLM"):
    print(chunk.text , end=" ", flush=True)
```

```python
for chunk in model.stream("Why do parrots have colourful feathers"):
    print(chunk.text , end=" ", flush=True)
```

```python
model.invoke("Why do parrots have colourful feathers?")
```

```python
response = model.batch([
    "Why parrots have colourful feathers?",
    "Why do parrots talk?", 
])
for response in response:
    print(response)
```

```python
response = model.batch([
    "Why parrots have colourful feathers?",
    "Why do parrots talk?", 
],
config={
    'max_concurrency': 5,
})
for response in response:
    print(response)
```
