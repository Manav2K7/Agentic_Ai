Model Integration With OpenAI , Google gemini , Groq


```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_GENAI_API_KEY"] = os.getenv("GOOGLE_GENAI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

Google gemini Integration

```python
import os
from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-3.6-flash")
response = model.invoke("Why do parrots talk?")
response.content
```

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
response = model.invoke("Why do parrots talk?") 
response.content 
```

```python
ChatopenAI

from langchain_openai import ChatOpenAI
model = ChatOpenAI(model_name="gpt-4.1")
response = model.invoke("Hello everyone")
response
```

GROQ model integration


```python
from langchain.agents import create_agent
from langchain_core.tools import tool

from langchain.chat_models import init_chat_model
model = init_chat_model("groq:llama-3.1-8b-instant")
response = model.invoke("Hello How are you?")   
response.content
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model_name="llama-3.1-8b-instant")
response = model.invoke("Hello How are you?")
response.content
```
