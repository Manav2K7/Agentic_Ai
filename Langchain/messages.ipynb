```python
import os

from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")  
model = init_chat_model("groq:llama-3.1-8b-instant")


```

```python
model.invoke("Please tell me what is Artificial Intelligence")
```

Text

```python
model.invoke("what is Langchain")
```

Tool Messages

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("You are a poetry expert."),
    HumanMessage("Write a poem about the Artificial intelligence."),
  
    
]
response = model.invoke(messages)
response.content
```

```python
system_msg = SystemMessage("You are a helpful assistant.")

messages =[
    system_msg,
    HumanMessage("How to create a Rest Api?")
]
response = model.invoke(messages)
print(response.content)
```

```python
system_msg = SystemMessage("You are a helpful coding assistant.")

messages =[
    system_msg,
    HumanMessage("How to create a Rest Api?")
]
response = model.invoke(messages)
print(response.content)
```

detailed info to the LLm


```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

system_msg = SystemMessage(
    """You are a senior python developer and a very experienced web framework developer. You have a lot of experience in creating Rest APIs using FastAPI and Django Rest Framework. You are also very good at writing clean and efficient code.
    Always provide code examples and explain reasoning .
    Be concise but though in tyour explanations."""
)


messages =[
    system_msg,
    HumanMessage("How to create a Rest Api?")
]
response = model.invoke(messages)
print(response.content)
```

```python
human_msg = HumanMessage(
    content = "Hello",
    name="Alice",
    id = "msg_123",
)
```

```python
response = model.invoke([ human_msg])
response
```

```python
from langchain.messages import HumanMessage, SystemMessage,AIMessage

ai_msg = AIMessage("i will be happy to help you with that question")

messages = [
    SystemMessage("You are a helpful assistant."),
    ai_msg,
    HumanMessage("Great! What is 2+2?"),
]

response = model.invoke(messages)
print(response.content)
```
